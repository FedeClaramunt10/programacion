1 - Definir una función lambda que reconozca si una secuencia de
caracteres es una variable válida en python (no comienza con un
número o puede comenzar con guión bajo)

```python
# Definimos una función lambda para verificar si una cadena es un nombre de variable válido en Python.
# Un nombre de variable válido no debe comenzar con un número y puede comenzar con guión bajo.

es_nombre_variable_valido = lambda s: s.isidentifier() and not s[0].isdigit()

# La función lambda es_nombre_variable_valido toma una cadena s como entrada y retorna True si es un nombre de variable válido según los requisitos, y False en caso contrario.

# s.isidentifier(): Este método verifica si la cadena es un identificador válido en Python, es decir, si contiene solo caracteres permitidos para nombres de variables (letras, números y guiones bajos) y no es una palabra clave reservada.
# not s[0].isdigit(): Esta parte adicional asegura que el primer carácter de la cadena no sea un dígito.

# ejemplos para verificar su funcionamiento:
print(f"'mi_variable': {es_nombre_variable_valido('mi_variable')}")
print(f"'_otra_variable': {es_nombre_variable_valido('_otra_variable')}")
print(f"'variable123': {es_nombre_variable_valido('variable123')}")
print(f"'123variable': {es_nombre_variable_valido('123variable')}")
print(f"'variable con espacios': {es_nombre_variable_valido('variable con espacios')}")
print(f"'variable-con-guion': {es_nombre_variable_valido('variable-con-guion')}")
```

2 - Dada una lista de tuplas donde cada tupla es (ciudad, población), usa
filter() para extraer las ciudades con una población mayor a 2
millones.


```python
lista_ciudades = [ ('Tokio', 37000000),
('Delhi', 28000000),
('Bogotá', 7000000),
('París', 2100000),
('Caracas', 1900000) ]

# Usamos filter() con una función lambda para extraer las ciudades con población mayor a 2 millones.
ciudades_millonarias = list(filter(lambda ciudad: ciudad[1] > 2000000, lista_ciudades))

# Mostramos el resultado
print("Ciudades con población mayor a 2 millones:")
print(ciudades_millonarias)
```

3. Dado un diccionario de personas (con claves como 'nombre' y 'edad'),
filtra aquellas personas que son menores de 18 años.

```python
lista_personas = [{'nombre': 'Ana', 'edad': 25},
{'nombre': 'Juan', 'edad': 15},
{'nombre': 'Sofía', 'edad': 30},
{'nombre': 'Pedro', 'edad': 17},
{'nombre': 'Elena', 'edad': 18} ]

# Usamos filter() con una función lambda para extraer las personas menores de 18 años.
personas_menores_18 = list(filter(lambda persona: persona['edad'] < 18, lista_personas))

# Mostramos el resultado
print("Personas menores de 18 años:")
print(personas_menores_18)
```

4. Ordena una lista de cadenas alfabéticamente, ignorando si las letras
están en mayúsculas o minúsculas.


```python
lista_cadenas = ["Banana", "apple", "Orange", "grape", "Kiwi"]

# Ordenamos la lista alfabéticamente, ignorando mayúsculas y minúsculas.
# Usamos el método sort() con key=str.lower
lista_cadenas.sort(key=str.lower)

# Mostramos el resultado
print("Lista de cadenas ordenada (ignorando mayúsculas/minúsculas):")
print(lista_cadenas)
```

5. Ordenar una lista de coches almacenados como tuplas

```python
tuplas_coches = [
('Rojo', '4859-A', 'A'),
('Azul', '2901-Z', 'M'),
('Gris', '1892-B', 'M') ]

# Ordenamos la lista de tuplas de coches por matrícula (el segundo elemento de la tupla)
# Usamos el método sort() con una función lambda para especificar la clave de ordenación
tuplas_coches.sort(key=lambda coche: coche[1])

# Mostramos el resultado
print("Lista de coches ordenada por matrícula:")
print(tuplas_coches)
```
