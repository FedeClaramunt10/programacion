### **Paso 1: Diseño de la clase base `Empleado` con validaciones básicas**

Empezamos con la clase `Empleado`. Esta será la 'madre' de todas las clases de empleados. Contendrá la información general que todo empleado tiene (nombre, apellido, fecha de ingreso) y definirá cómo debe ser la estructura para calcular el salario y mostrar la liquidación.

**¡Importante!** Para cumplir con los criterios de evaluación, vamos a añadir algunas validaciones básicas en el constructor (`__init__`). Por ejemplo, asegurarnos de que el nombre y el apellido no estén vacíos y que la fecha de ingreso sea un objeto de tipo `date`.

```python
# Importamos el módulo date para manejar fechas de forma correcta.
from datetime import date

class Empleado:
    """Clase base que representa a cualquier empleado de la empresa.
    Contiene atributos comunes y define la estructura de métodos como calcular_salario y mostrar_liquidacion.
    """

    def __init__(self, nombre: str, apellido: str, fecha_ingreso: date):
        """Constructor de la clase Empleado.

        Args:
            nombre (str): El nombre del empleado. No puede estar vacío.
            apellido (str): El apellido del empleado. No puede estar vacío.
            fecha_ingreso (date): La fecha en que el empleado ingresó a la empresa. Debe ser un objeto date.

        Validaciones de datos:
        * Se asegura que el nombre y apellido no sean cadenas vacías.
        * Se asegura que fecha_ingreso sea una instancia válida de datetime.date.
        """
        # Validación para el nombre
        if not nombre or not isinstance(nombre, str):
            raise ValueError("El nombre del empleado no puede estar vacío y debe ser una cadena de texto.")
        self.nombre = nombre

        # Validación para el apellido
        if not apellido or not isinstance(apellido, str):
            raise ValueError("El apellido del empleado no puede estar vacío y debe ser una cadena de texto.")
        self.apellido = apellido

        # Validación para la fecha de ingreso
        if not isinstance(fecha_ingreso, date):
            raise TypeError("La fecha de ingreso debe ser un objeto de tipo datetime.date.")
        self.fecha_ingreso = fecha_ingreso

        print(f"Empleado creado: {self.nombre} {self.apellido}") # Mensaje para saber que se creó correctamente.

    def calcular_salario(self):
        """Método que **debe ser implementado** por las clases hijas (subclases).

        En la clase `Empleado`, este método no tiene una implementación real porque el cálculo del salario
        depende del tipo de empleado. Lanzamos un `NotImplementedError` para forzar que las subclases lo implementen.
        """
        raise NotImplementedError("El método 'calcular_salario' debe ser implementado por las subclases (EmpleadoTiempoCompleto, EmpleadoPorHoras).")

    def mostrar_liquidacion(self):
        """Imprime los datos del empleado y su salario calculado.

        Llama al método `calcular_salario()` (que será el de la subclase) para obtener el monto.
        """
        try:
            salario = self.calcular_salario()
            print(f"\n--- Liquidación de Empleado --- ")
            print(f"Nombre Completo: {self.nombre} {self.apellido}")
            print(f"Fecha de Ingreso: {self.fecha_ingreso.strftime('%d-%m-%Y')}") # Formateamos la fecha a un formato legible.
            print(f"Salario Calculado: ${salario:,.2f}") # Formateamos el salario con 2 decimales para que se vea bonito.
        except NotImplementedError as e:
            print(f"Error al mostrar liquidación: {e}")
        except Exception as e:
            print(f"Ocurrió un error inesperado al calcular la liquidación: {e}")
```

### **Paso 2: Creación de clases específicas de empleados con sus propias validaciones**

Ahora, vamos a crear las dos clases que heredan de `Empleado`. Esto significa que `EmpleadoTiempoCompleto` y `EmpleadoPorHoras` tendrán automáticamente los atributos y métodos de `Empleado`, pero cada una implementará su propia lógica para `calcular_salario()` y tendrá atributos específicos. Añadiremos validaciones para sus atributos propios.

```python
class EmpleadoTiempoCompleto(Empleado):
    """Representa a un empleado que recibe un salario mensual fijo. Hereda de Empleado."""

    def __init__(self, nombre: str, apellido: str, fecha_ingreso: date, salario_mensual: float):
        """Constructor de EmpleadoTiempoCompleto.

        Args:
            nombre (str): El nombre del empleado.
            apellido (str): El apellido del empleado.
            fecha_ingreso (date): La fecha de ingreso.
            salario_mensual (float): El salario fijo que el empleado recibe cada mes.

        Validaciones de datos:
        * Se asegura que el salario_mensual sea un número positivo.
        """
        # Primero, llamamos al constructor de la clase base `Empleado` para inicializar sus atributos.
        # Esto es importante para reusar el código y las validaciones de `Empleado`.
        super().__init__(nombre, apellido, fecha_ingreso)

        # Luego, validamos e inicializamos el atributo específico de esta clase.
        if not isinstance(salario_mensual, (int, float)) or salario_mensual <= 0:
            raise ValueError("El salario mensual debe ser un número positivo.")
        self.salario_mensual = salario_mensual

        print(f"  -> Tipo: Tiempo Completo (Salario: ${self.salario_mensual:,.2f})")

    def calcular_salario(self):
        """Implementación del método `calcular_salario()` para EmpleadoTiempoCompleto.

        Para este tipo de empleado, el salario es simplemente el `salario_mensual`.
        """
        return self.salario_mensual


class EmpleadoPorHoras(Empleado):
    """Representa a un empleado al que se le paga por las horas trabajadas. Hereda de Empleado."""

    def __init__(self, nombre: str, apellido: str, fecha_ingreso: date, horas_trabajadas: float, tarifa_por_hora: float):
        """Constructor de EmpleadoPorHoras.

        Args:
            nombre (str): El nombre del empleado.
            apellido (str): El apellido del empleado.
            fecha_ingreso (date): La fecha de ingreso.
            horas_trabajadas (float): La cantidad de horas que el empleado trabajó.
            tarifa_por_hora (float): El precio por cada hora trabajada.

        Validaciones de datos:
        * Se asegura que horas_trabajadas y tarifa_por_hora sean números positivos.
        """
        # Llamamos al constructor de la clase base `Empleado`.
        super().__init__(nombre, apellido, fecha_ingreso)

        # Validamos e inicializamos los atributos específicos de esta clase.
        if not isinstance(horas_trabajadas, (int, float)) or horas_trabajadas < 0:
            raise ValueError("Las horas trabajadas deben ser un número positivo o cero.")
        self.horas_trabajadas = horas_trabajadas

        if not isinstance(tarifa_por_hora, (int, float)) or tarifa_por_hora <= 0:
            raise ValueError("La tarifa por hora debe ser un número positivo.")
        self.tarifa_por_hora = tarifa_por_hora

        print(f"  -> Tipo: Por Horas (Horas: {self.horas_trabajadas}, Tarifa: ${self.tarifa_por_hora:,.2f})")

    def calcular_salario(self):
        """Implementación del método `calcular_salario()` para EmpleadoPorHoras.

        El salario se calcula multiplicando las `horas_trabajadas` por la `tarifa_por_hora`.
        """
        return self.horas_trabajadas * self.tarifa_por_hora
```

**Ejercicio 2: Sistema de Alquiler de Vehículos**

### **Paso 1: Diseño de la clase `Motor`**

Primero, vamos a crear la clase `Motor`. Esta clase será bastante sencilla, solo almacenará la cilindrada y el tipo de combustible del motor de un vehículo. También tendrá un método para describir el motor.

```python
class Motor:
    """Clase que representa el motor de un vehículo."""

    def __init__(self, cilindrada: str, tipo_combustible: str):
        """Constructor de la clase Motor.

        Args:
            cilindrada (str): La cilindrada del motor (ej. "2.0L").
            tipo_combustible (str): El tipo de combustible que usa el motor (ej. "Gasolina", "Diésel").
        """
        # Asignamos los atributos al crear una instancia de Motor.
        self.cilindrada = cilindrada
        self.tipo_combustible = tipo_combustible
        print(f"  -> Motor creado: {self.describir_motor()}")

    def describir_motor(self) -> str:
        """Devuelve una cadena de texto describiendo las características del motor.

        Returns:
            str: Descripción del motor.
        """
        # Formateamos la descripción para que sea fácil de leer.
        return f"Motor {self.cilindrada}, {self.tipo_combustible}"
```

### **Paso 2: Diseño de la clase `Vehiculo` con validaciones de kilometraje**

Ahora viene la clase `Vehiculo`, que será un poco más compleja. Tendrá atributos como la marca, el modelo, el kilometraje y, muy importante, una instancia de la clase `Motor` que acabamos de crear (esto es **composición**, donde una clase 'tiene' otra).

Lo más interesante aquí es la gestión del `kilometraje`. Necesitamos asegurarnos de que el kilometraje nunca disminuya y que siempre sea un valor no negativo. Para lograr esto, utilizaremos propiedades de Python (`@property` y `@kilometraje.setter`) para controlar cómo se accede y se modifica este atributo.

```python
class Vehiculo:
    """Clase que representa un vehículo en el sistema de alquiler."""

    def __init__(self, marca: str, modelo: str, kilometraje_inicial: float, motor: Motor):
        """Constructor de la clase Vehiculo.

        Args:
            marca (str): La marca del vehículo.
            modelo (str): El modelo del vehículo.
            kilometraje_inicial (float): El kilometraje con el que inicia el vehículo.
            motor (Motor): Una instancia de la clase Motor asociada a este vehículo.
        """
        self.marca = marca
        self.modelo = modelo
        self._kilometraje = 0.0  # Inicializamos con 0, el setter validará el valor inicial.
        self.kilometraje = kilometraje_inicial # Usamos el setter para la validación inicial.
        self.motor = motor # Aquí 'componemos' el vehículo con una instancia de Motor.
        print(f"Vehículo creado: {self.marca} {self.modelo} (Kilometraje inicial: {self._kilometraje} km)")

    @property
    def kilometraje(self) -> float:
        """Getter para el atributo kilometraje.
        Permite acceder al kilometraje como vehiculo.kilometraje.
        """
        return self._kilometraje

    @kilometraje.setter
    def kilometraje(self, nuevo_kilometraje: float):
        """Setter para el atributo kilometraje.
        Aplica las reglas de validación cada vez que se intenta modificar el kilometraje.

        Args:
            nuevo_kilometraje (float): El nuevo valor a intentar asignar al kilometraje.
        """
        # Regla 1: El nuevo valor no puede ser negativo.
        if nuevo_kilometraje < 0:
            print(f"[ERROR] El kilometraje no puede ser negativo. Se intentó asignar: {nuevo_kilometraje}")
            return # No cambiamos el valor si falla la validación.

        # Regla 2: El nuevo valor no puede ser menor al valor que ya tiene.
        if nuevo_kilometraje < self._kilometraje:
            print(f"[ERROR] El kilometraje no puede disminuir. Actual: {self._kilometraje} km, Intentado: {nuevo_kilometraje} km.")
            return # No cambiamos el valor si falla la validación.

        # Si todas las validaciones pasan, actualizamos el kilometraje.
        self._kilometraje = nuevo_kilometraje
        # print(f"Kilometraje actualizado a: {self._kilometraje} km") # Opcional: mensaje de éxito.

    def recorrer_distancia(self, km: float):
        """Incrementa el kilometraje del vehículo.

        Esta acción usa el setter de kilometraje para aplicar las mismas reglas de validación.

        Args:
            km (float): La distancia en kilómetros a recorrer.
        """
        if km <= 0:
            print(f"[ERROR] La distancia a recorrer debe ser un valor positivo. Se intentó: {km} km.")
            return

        # Intentamos actualizar el kilometraje usando el setter, que ya tiene las validaciones.
        print(f"Intentando recorrer {km} km...")
        self.kilometraje = self._kilometraje + km

    def obtener_descripcion(self) -> str:
        """Devuelve una cadena con las características completas del vehículo, incluyendo el motor.

        Returns:
            str: Descripción detallada del vehículo.
        """
        # Usamos el método describir_motor() del objeto Motor para obtener sus detalles.
        return f"Vehículo: {self.marca} {self.modelo} ({self.kilometraje} km) - {self.motor.describir_motor()}"
```

### **Paso 3: Demostración del Sistema de Alquiler de Vehículos**

Ahora que tenemos ambas clases listas y la clase `Vehiculo` con su lógica de validación de kilometraje, vamos a realizar las pruebas que nos pide el ejercicio para ver cómo funciona todo.

```python
# --- DEMOSTRACIÓN DEL SISTEMA DE ALQUILER DE VEHÍCULOS ---
print("\n--- Iniciando demostración del Sistema de Alquiler de Vehículos ---")

# 1. Crear una instancia de Motor
print("\nCreando un motor...")
mi_motor = Motor(cilindrada="1.8L", tipo_combustible="Híbrido")

# 2. Crear una instancia de Vehiculo usando el motor anterior
print("\nCreando un vehículo...")
mi_vehiculo = Vehiculo(
    marca="Toyota",
    modelo="Corolla",
    kilometraje_inicial=15000.0,
    motor=mi_motor # Le pasamos la instancia de Motor que acabamos de crear.
)

# 3. Mostrar la descripción del vehículo.
print("\nDescripción inicial del vehículo:")
print(mi_vehiculo.obtener_descripcion())

# 4. Intentar asignar un kilometraje menor al actual y verificar que el sistema lo impide.
print("\nIntentando asignar un kilometraje menor (5000 km)...")
mi_vehiculo.kilometraje = 5000.0 # Esto debería fallar por la validación.
print(f"Kilometraje actual después del intento fallido: {mi_vehiculo.kilometraje} km")

print("\nIntentando asignar un kilometraje negativo (-100 km)...")
mi_vehiculo.kilometraje = -100.0 # Esto también debería fallar.
print(f"Kilometraje actual después del intento fallido: {mi_vehiculo.kilometraje} km")

# Intentamos asignar un kilometraje válido pero igual al actual.
print("\nIntentando asignar el mismo kilometraje (15000 km)...")
mi_vehiculo.kilometraje = 15000.0 # Esto debería ser permitido pero sin cambio.
print(f"Kilometraje actual después del intento: {mi_vehiculo.kilometraje} km")

# 5. Usar el método recorrer_distancia(150) y mostrar la descripción nuevamente.
print("\nRecorriendo una distancia de 150 km...")
mi_vehiculo.recorrer_distancia(150.0)
print("\nDescripción del vehículo después de recorrer la distancia:")
print(mi_vehiculo.obtener_descripcion())

# Intentamos recorrer una distancia negativa (debería fallar)
print("\nIntentando recorrer una distancia negativa (-50 km)...")
mi_vehiculo.recorrer_distancia(-50.0)
print("\nDescripción del vehículo después del intento fallido:")
print(mi_vehiculo.obtener_descripcion())

print("\n--- Demostración completada con éxito. ---")
```

**Ejercicio 3: Carrito de Compras**

### **Paso 1: Diseño de la clase base `Producto`**

Comenzamos con la clase `Producto`. Esta será la clase fundamental para cualquier artículo que pueda ser vendido. Tendrá un **SKU** (Stock Keeping Unit, un identificador único) y un **precio base**. Definiremos un método `calcular_precio_final` que, al igual que en `Empleado`, será implementado de forma específica por sus subclases.

```python
class Producto:
    """Clase base para representar un producto genérico en el sistema de carrito de compras."""

    def __init__(self, sku: str, precio_base: float):
        """Constructor de la clase Producto.

        Args:
            sku (str): Identificador único del producto (SKU).
            precio_base (float): El precio base del producto antes de impuestos.

        Validaciones:
        * El SKU no puede estar vacío.
        * El precio base debe ser un número positivo.
        """
        if not sku or not isinstance(sku, str):
            raise ValueError("El SKU del producto no puede estar vacío y debe ser una cadena de texto.")
        self.sku = sku

        if not isinstance(precio_base, (int, float)) or precio_base <= 0:
            raise ValueError("El precio base debe ser un número positivo.")
        self.precio_base = precio_base

        print(f"Producto base creado: {self.sku} (Precio Base: ${self.precio_base:,.2f})")

    def calcular_precio_final(self) -> float:
        """Método que **debe ser implementado** por las subclases.

        Este método es abstracto en la clase base, ya que el cálculo del IVA/precio final
        depende del tipo específico de producto.
        """
        raise NotImplementedError("El método 'calcular_precio_final' debe ser implementado por las subclases.")
```

### **Paso 2: Creación de clases de productos específicos**

Ahora vamos a crear las clases `ProductoAlimenticio` y `ProductoElectronico`. Ambas heredarán de `Producto` y añadirán sus propios atributos (fecha de caducidad, garantía en meses) y, crucialmente, implementarán el método `calcular_precio_final()` con su lógica de IVA correspondiente.

```python
from datetime import date

class ProductoAlimenticio(Producto):
    """Representa un producto alimenticio, que hereda de Producto."""

    def __init__(self, sku: str, precio_base: float, fecha_caducidad: date):
        """Constructor de ProductoAlimenticio.

        Args:
            sku (str): Identificador único del producto.
            precio_base (float): Precio base del producto.
            fecha_caducidad (date): La fecha de caducidad del producto.

        Validaciones:
        * La fecha de caducidad debe ser un objeto `date`.
        """
        super().__init__(sku, precio_base)
        if not isinstance(fecha_caducidad, date):
            raise TypeError("La fecha de caducidad debe ser un objeto de tipo datetime.date.")
        self.fecha_caducidad = fecha_caducidad
        print(f"  -> Producto Alimenticio: {self.sku} (Caducidad: {self.fecha_caducidad.strftime('%d-%m-%Y')})")

    def calcular_precio_final(self) -> float:
        """Calcula el precio final para un producto alimenticio (precio_base + 5% IVA)."""
        iva = 0.05
        return self.precio_base * (1 + iva)


class ProductoElectronico(Producto):
    """Representa un producto electrónico, que hereda de Producto."""

    def __init__(self, sku: str, precio_base: float, garantia_meses: int):
        """Constructor de ProductoElectronico.

        Args:
            sku (str): Identificador único del producto.
            precio_base (float): Precio base del producto.
            garantia_meses (int): La duración de la garantía en meses.

        Validaciones:
        * La garantía en meses debe ser un entero no negativo.
        """
        super().__init__(sku, precio_base)
        if not isinstance(garantia_meses, int) or garantia_meses < 0:
            raise ValueError("La garantía en meses debe ser un número entero no negativo.")
        self.garantia_meses = garantia_meses
        print(f"  -> Producto Electrónico: {self.sku} (Garantía: {self.garantia_meses} meses)")

    def calcular_precio_final(self) -> float:
        """Calcula el precio final para un producto electrónico (precio_base + 15% IVA)."""
        iva = 0.15
        return self.precio_base * (1 + iva)
```

### **Paso 3: Creación de la clase `CarritoDeCompras`**

Ahora construiremos la clase `CarritoDeCompras`. Esta clase 'contendrá' una lista de productos y se encargará de gestionar las operaciones del carrito: agregar productos, calcular el total de la compra y mostrar un ticket detallado.

```python
class CarritoDeCompras:
    """Clase que representa un carrito de compras, gestionando una colección de productos."""

    def __init__(self):
        """Constructor de CarritoDeCompras.
        Inicializa una lista vacía para almacenar los productos.
        """
        self.productos = [] # Aquí guardaremos todos los productos que se añadan al carrito.
        print("Carrito de compras creado.")

    def agregar_producto(self, producto: Producto):
        """Agrega un producto al carrito de compras.

        Args:
            producto (Producto): Una instancia de Producto (o sus subclases) a añadir.
        """
        if not isinstance(producto, Producto):
            raise TypeError("Solo se pueden agregar instancias de Producto o sus subclases al carrito.")
        self.productos.append(producto)
        print(f"Producto '{producto.sku}' agregado al carrito.")

    def calcular_total(self) -> float:
        """Calcula la suma de los precios finales de todos los productos en el carrito.

        Returns:
            float: El precio total de la compra.
        """
        total = 0.0
        for producto in self.productos:
            total += producto.calcular_precio_final() # Aquí aprovechamos el polimorfismo!
        return total

    def mostrar_ticket(self):
        """Imprime un ticket detallado con cada producto y el total de la compra."""
        print("\n--- TICKET DE COMPRA ---")
        if not self.productos:
            print("El carrito está vacío.")
            return

        for producto in self.productos:
            # Obtenemos el precio final de cada producto, que ya incluye el IVA específico.
            precio_final = producto.calcular_precio_final()
            print(f"  SKU: {producto.sku} - Precio Final: ${precio_final:,.2f}")

        # Calculamos y mostramos el total de la compra.
        total_compra = self.calcular_total()
        print("------------------------")
        print(f"TOTAL A PAGAR: ${total_compra:,.2f}")
        print("------------------------")
```

### **Paso 4: Demostración - Llenando un carrito de compras**

Finalmente, vamos a crear algunos productos, los agregaremos a un carrito de compras y luego mostraremos el ticket para ver cómo todo nuestro sistema trabaja en conjunto.
